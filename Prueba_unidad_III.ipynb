# Ejercicio 1: Control de Tren (Valor - 4 pts)

### Descripción
Simula el sistema de control de un tren que debe responder a señales en la vía. Las señales pueden ser **"verde"**, **"amarillo"** o **"rojo"**. El tren debe ajustar su velocidad base según la señal actual y la d hasta la próxima estación: 

- **Señal verde**: 
  - Acelera a **100 km/h** si la d es mayor a 5 km. 
  - Mantiene **80 km/h** en caso contrario. 
- **Señal amarilla**: 
  - Reduce la v a **50 km/h** si la d es mayor a 3 km. 
  - De lo contrario, reduce a **30 km/h**. 
- **Señal roja**: 
  - Frena hasta **0 km/h** inmediatamente. 
- **Obstáculo en la vía**: 
  - El tren debe frenar a **0 km/h**, independientemente de la señal. 

--- 
### ✅ Salida esperada 
El programa debe devolver la v actual del tren en base a las condiciones dadas. 

**Ejemplo:** 
- Entrada: `señal="amarillo", d=4, obstáculo=False`
- Salida: `v: 50 km/h`

### Restricciones

- Debe utilizar estructuras condicionales.
- No se permite crear listas ni funciones.

````python
# Aqui se realiza el ejercicio

#INGRESO DE DATOS
s = str(input("Cual es la Señal (verde, amarillo, rojo) :")).lower()
d = float(input("Cual es la d en Km:"))
o = bool(input("Hay obstaculos Si(True) No(False)):"))

v = 0 #VARIABLE COMO DE ESTADO O BANDERA PARA GUARDAR LA VELOCIDAD 

if o == "si": # AQUI VERIFICO SI HAY OBSTACULOS O NO 
    v = 0
else:
    # Evaluamos las señales solo si la vía está libre
    if s == "verde":
        if d > 5:
            v = 100
        else:
            v = 80
    elif s == "amarillo":
        if d > 3:
            v = 50
        else:
            v = 30
    elif s == "rojo":
        v = 0
    else:
        print("ERROR")

print(f"v: {v} Km/h")
````

# Ejercicio 2: Buscador del Próximo Año Bisiesto (Valor - 4 pts)

### Descripción
Una empresa de planificación necesita conocer cuál será el próximo año bisiesto después de un año dado.

Un año es bisiesto cuando:

- Es divisible entre 4 y no entre 100.
- O es divisible entre 400.

El programa debe solicitar un año y utilizar un ciclo para encontrar el siguiente año bisiesto.

### ✅ Salida esperada

**Ejemplo:**
- Entrada: `2025`  
- Salida: `El próximo año bisiesto es: 2028`
____________

- Entrada: `2096`  
- Salida: `El próximo año bisiesto es: 2104`

### Restricciones

- Debe utilizar un ciclo **while** para encontrar el próximo año bisiesto.
- Debe utilizar estructuras condicionales.
- No se permite ingresar varios años.

````python
# Aqui se resuelve el ejercicio

# y = an;o para no usar la ñ por que el teclado no medeja ponerla y me pone ; y de Year en ingles profe 

y = int(input("Ingrese un año: "))

#  como dice que es el proximo a;o le sumaos uno 
siguienteY = y + 1
encontrado = False

# Ciclo while para seguir buscando hasta encontrar el bisiesto
while encontrado == False:
    # Veo si el año actual cumple las reglas de bisiesto dadas de divisible entre 4 y no entre 100
    if (siguienteY % 4 == 0 and siguienteY % 100 != 0) or (siguienteY % 400 == 0):
        encontrado = True
    else:
        # Si no es bisiesto, pasamos al siguiente año
        siguienteY += 1

print(f"El próximo año Bisiesto es: {siguienteY}")
````

# Ejercicio 3: Secuencia Misteriosa (Valor - 5 pts)

### Descripción
Un laboratorio matemático está estudiando una secuencia numérica especial.

Dado un número entero positivo:

- Si el número es par, se divide enteramente entre 2.
- Si el número es impar, se multiplica por 3 y se suma 1.

El proceso se repite hasta que el número llegue a 1.

### ✅ Salida esperada
El programa debe mostrar cada valor generado y contar cuántos pasos fueron necesarios.

**Ejemplo:**
- Entrada: `12`  
- Salida: `Secuencia: 12, 6, 3, 10, 5, 16, 8, 4, 2, 1; Pasos: 9`
___

- Entrada: `45`  
- Salida: `Secuencia: 45, 136, 68, 34, 17, 52, 26, 13, 40, 20, 10, 5, 16, 8, 4, 2, 1; Pasos: 16`

### Restricciones

- Debe utilizar un ciclo while.
- Debe utilizar estructuras condicionales.
- El número ingresado debe ser mayor que cero.

````python
# Aqui se resulve el ejercicio

n = int(input("Ingrese un número entero positivo: "))

# Verificamos que el número cumpla la restricción
if n > 0:
    secuencia = str(n)
    pasos = 0
    
    # El ciclo se repite hasta que el número llegue a 1
    while n != 1:
        # Si es par
        if n % 2 == 0:
            n = n // 2
        # Si es impar
        else:
            n = (n * 3) + 1
            
        # Agregamos el nuevo número a nuestra secuencia y sumamos un paso
        secuencia += ", " + str(n)
        pasos += 1
        
    print(f"Secuencia: {secuencia}; Pasos: {pasos}")
else:
    print("Error: El número debe ser mayor que cero.")
````

# Ejercicio 4: Centro de Control de Robots (Valor - 7 pts)

### Descripción
Un centro de automatización controla varios robots industriales.

- El usuario indicará cuántos robots desea procesar.
- Para cada robot se solicitará su estado actual:

    - `"activo"`
    - `"espera"`
    - `"mantenimiento"`
    - `"apagado"`

El programa debe utilizar la **estructura match** para determinar la acción correspondiente:

| Estado        | Acción                  |
|---------------|-------------------------|
| Activo        | Operando normalmente    |
| Espera        | Esperando instrucciones |
| Mantenimiento | Revisar componentes     |
| Apagado       | Sistema detenido        |

Además:
- Debe llevar un conteo de cuántos robots se encuentran en cada estado.
- Al finalizar, mostrar un resumen general.
---

### ✅ Salida esperada

El programa debe mostrar la acción correspondiente para cada robot procesado y, al finalizar, presentar un resumen con la cantidad de robots en cada estado.

**Entrada**

```text
Cantidad de robots: 4

activo
espera
activo
apagado
```
___
**Salida**

```text
Robot 1: Operando normalmente
Robot 2: Esperando instrucciones
Robot 3: Operando normalmente
Robot 4: Sistema detenido

Resumen:
Activo: 2
Espera: 1
Mantenimiento: 0
Apagado: 1
```
___
### Restricciones

- Debe utilizar un ciclo for.
- Debe utilizar la estructura match.
- No se permite utilizar listas.

````python
# Aquí se resuelve el ejercicio


cantidadRobot = int(input("Cantidad de robots: "))

# Variables para llevar el conteo de cada estado
a = 0
e = 0
m = 0
apagados = 0


print("-----------------------------------------") 

# Ciclo para  ver cada  robotcito 
for i in range(1, cantidadRobot + 1):
    estado = input(f"Estado del robot {i}: ").strip().lower()
    
    #  determino cada accion  
    match estado:
        case "activo":

            print(f"Robot {i}: Operando normalmente")
            a += 1

        case "espera":
            print(f"Robot {i}: Esperando instrucciones")
            e += 1

        case "mantenimiento":
            print(f"Robot {i}: Revisar componentes")
            m += 1

        case "apagado":
            print(f"Robot {i}: Sistema detenido")
            apagados += 1

        case _:
            print("ERROR")
            print(f"Robot {i}: Estado no válido")


print("-----------------------------------------") 

print("Resumen:")
print(f"Activo: {a}")
print(f"Espera: {e}")
print(f"Mantenimiento: {m}")
print(f"Apagado: {apagados}")

print("-----------------------------------------") 
````
